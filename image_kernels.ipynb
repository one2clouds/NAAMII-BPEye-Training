```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('pic.png', cv2.IMREAD_GRAYSCALE)

# Sobel edge detection kernels
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

sobel_y = np.array([[-1, -2, -1],
                    [0, 0, 0],
                    [1, 2, 1]])

# Apply the Sobel filters using cv2.filter2D
edges_x = cv2.filter2D(image, -1, sobel_x)
edges_y = cv2.filter2D(image, -1, sobel_y)

# Combine the two directions
edges = np.sqrt(edges_x**2 + edges_y**2)
edges = np.uint8(edges)

# Display the result
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.show()


plt.imshow(edges_x, cmap='gray')
plt.title('Edges in X direction')
plt.show()


plt.imshow(edges_y, cmap='gray')
plt.title('Edges in Y direction')
plt.show()


plt.imshow(edges, cmap='gray')
plt.title('Combined Edges')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread('pic.png', cv2.IMREAD_GRAYSCALE)

# Define a threshold value (between 0 and 255)
threshold_value = 128

# Apply binary thresholding using cv2.threshold
_, thresholded_image = cv2.threshold(image, threshold_value, 255, cv2.THRESH_BINARY)

# Display the original and thresholded images
plt.figure(figsize=(10, 5))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Thresholded image
plt.subplot(1, 2, 2)
plt.imshow(thresholded_image, cmap='gray')
plt.title('Thresholded Image')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np

img = cv2.imread('pic.png')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

gray_float = np.float32(gray)
dst = cv2.cornerHarris(gray_float, blockSize=2, ksize=3, k=0.04)

# Dilate the corner image to enhance corner points
dst = cv2.dilate(dst, None)

# Iterate through the corner matrix and mark red points on corners
img[dst > 0.01 * dst.max()] = [0, 0, 255]
cv2.imwrite('harris-corners-image.jpg', img)
```
