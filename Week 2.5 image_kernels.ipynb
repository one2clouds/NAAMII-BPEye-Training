```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(r"/home/s_ach/Desktop/EyePACS-DEV-RG-1.jpg", cv2.IMREAD_GRAYSCALE)

sobel_x = np.array([[-1, 0, 1],
                    [-1, 0, 1],
                    [-1, 0, 1]])

sobel_y = np.array([[-1, -2, -1],
                    [0, 0, 0],
                    [1, 2, 1]])

edges_x = cv2.filter2D(image, -1, sobel_x)
edges_y = cv2.filter2D(image, -1, sobel_y)

edges = edges_x + edges_y
edges = np.uint8(edges)

plt.imshow(edges_y, cmap='gray')
plt.title('Edges in Y direction')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(r"/home/s_ach/Desktop/EyePACS-DEV-RG-1.jpg", cv2.IMREAD_GRAYSCALE)
threshold_value = 120
_, thresholded_image = cv2.threshold(image, threshold_value, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(thresholded_image, cmap='gray')
plt.title('Thresholded Image')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np

img = cv2.imread('/home/s-ach/Desktop/EyePACS-TRAIN-RG-2891.jpg')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_float = np.float32(gray)
dst = cv2.cornerHarris(gray_float, blockSize=2, ksize=3, k=0.04)

# Dilate the corner image to enhance corner points
dst = cv2.dilate(dst, None)

# Iterate through the corner matrix and mark points on corners
img[dst > 0.01 * dst.max()] = [0, 255, 0]
cv2.imwrite('harris-corners-image.jpg', img)

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```
